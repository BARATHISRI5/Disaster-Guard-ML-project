# DisasterGuard — LSTM Training

Training an LSTM time-series model for flood risk forecasting.

**Architecture:** LSTM(128) → LSTM(64) → Dense(32) → Dense(3, Sigmoid)
**Input:** 14 days × 6 features (rainfall, soil moisture, wind, river level, NDWI, temp)
**Output:** 24h, 48h, 72h flood probabilities

```python
import numpy as np
import sys
sys.path.append('../backend')

from models.lstm_model import (
    build_lstm_model, generate_training_data,
    train_lstm, FloodTimeSeriesDataset, predict_flood_timeseries
)

# Generate training data
X, y = generate_training_data(num_samples=1000)
print(f'Sequences: {X.shape}, Labels: {y.shape}')
```

```python
# Split and create datasets
split = int(0.8 * len(X))
train_ds = FloodTimeSeriesDataset(X[:split], y[:split])
val_ds = FloodTimeSeriesDataset(X[split:], y[split:])

print(f'Train: {len(train_ds)}, Val: {len(val_ds)}')
```

```python
# Build and train model
model = build_lstm_model()
results = train_lstm(model, train_ds, val_ds, epochs=30)
print(f'Best validation loss: {results["best_val_loss"]:.4f}')
```

```python
# Test prediction
test_seq = X[0]
prediction = predict_flood_timeseries(model, test_seq)
print(f'Flood risk predictions:')
for window, prob in prediction.items():
    print(f'  {window}: {prob:.1%}')
```
