# DisasterGuard — CNN Training

Training a binary CNN classifier for flood detection from satellite image patches.

**Architecture:** 4× Conv2D → MaxPool → Dropout(0.3) → Dense(256) → Sigmoid
**Input:** 256×256×3 RGB patches
**Output:** Flood probability 0-1

```python
import numpy as np
import sys
sys.path.append('../backend')

from models.cnn_model import build_cnn_model, train_model, evaluate_model
from data.fetch_sentinel import generate_mock_patches

# Generate training data
patches, labels = generate_mock_patches(num_patches=200, num_flood=100)
print(f'Dataset: {patches.shape}, Labels: {labels.shape}')
print(f'Flood: {labels.sum()}, No-flood: {len(labels) - labels.sum()}')
```

```python
# Split data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(patches, labels, test_size=0.2, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

print(f'Train: {X_train.shape}, Val: {X_val.shape}, Test: {X_test.shape}')
```

```python
# Build and train model
model = build_cnn_model(input_shape=(256, 256, 3))
history = train_model(model, X_train, y_train, X_val, y_val, epochs=20)
```

```python
# Evaluate
results = evaluate_model(model, X_test, y_test)
print(f'Test Accuracy: {results["accuracy"]:.2%}')
```
