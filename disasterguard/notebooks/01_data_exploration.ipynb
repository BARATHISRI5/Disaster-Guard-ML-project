# DisasterGuard — Data Exploration

Exploring satellite data from Sentinel-1 SAR, Sentinel-2 Optical, and MODIS
for Tamil Nadu flood detection.

**Target Districts:** Chennai, Thoothukudi, Nagapattinam, Tirunelveli
**Date Range:** 2015-2024

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append('../backend')

from data.fetch_sentinel import generate_mock_patches, DISTRICT_BOUNDS, FLOOD_EVENTS
from data.preprocess import calculate_ndwi, generate_mock_ndwi_timeseries

print('DisasterGuard Data Exploration')
print(f'Target districts: {list(DISTRICT_BOUNDS.keys())}')
```

```python
# Generate mock patches for visualization
patches, labels = generate_mock_patches(num_patches=20, num_flood=10)

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for i, ax in enumerate(axes[0]):
    ax.imshow(patches[i])
    ax.set_title(f'Flood (label={labels[i]})', fontsize=10)
    ax.axis('off')
for i, ax in enumerate(axes[1]):
    ax.imshow(patches[i+10])
    ax.set_title(f'No Flood (label={labels[i+10]})', fontsize=10)
    ax.axis('off')
plt.suptitle('Sample Satellite Patches', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# NDWI time series during flood event
ndwi_ts = generate_mock_ndwi_timeseries(num_days=30, flood_start=20)

plt.figure(figsize=(12, 4))
plt.plot(ndwi_ts, 'b-o', markersize=4)
plt.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
plt.axvspan(20, 30, alpha=0.2, color='red', label='Flood period')
plt.xlabel('Day')
plt.ylabel('NDWI')
plt.title('NDWI Time Series — Flood Detection')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
